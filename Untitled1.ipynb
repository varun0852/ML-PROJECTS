```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv("Advertising_Data.csv")
df
```

```python
df.info()
```

```python
df.describe()
```

```python
features = df[["TV", "Billboards", "Google_Ads", "Social_Media", "Influencer_Marketing", "Affiliate_Marketing"]]
```

```python
sns.boxplot(features)
```

```python
q1 = features["TV"].quantile(0.25)
q3 = features["TV"].quantile(0.75)
iqr = q3-q1
lower_fence = q1 - 1.5*iqr
upper_fence = q3 + 1.5*iqr
trimmed_features = features[(features['TV'] > lower_fence) & (features['TV'] < upper_fence)]
```

```python
trimmed_features
```

```python
negative_counts = 0
for i in features["TV"]:
    if i<0:
        negative_counts +=1
print(negative_counts)
```

```python
label = df["Product_Sold"]
```

```python
features
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(features, label, test_size=0.2)
```

```python
algorithm = LinearRegression()
```

```python
algorithm.fit(xtrain, ytrain)
```

```python
algorithm.coef_
```

```python
algorithm.predict([[770.05,	501.36,	694.60,	172.26,	572.26, 410.56]])
```

```python
ypred = algorithm.predict(xtest)
ypred
```

```python
r2_score(ypred, ytest)
```
